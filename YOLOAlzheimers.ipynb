```python
import gradio as gr
from ultralytics import YOLO
import numpy as np

# Load the trained model
model = YOLO('/content/runs/classify/train/weights/last.pt')

# Define a function to handle the prediction
def classify_image(image):
    # Perform inference
    results = model(image)

    # Get the class names and probabilities
    names_dict = results[0].names
    probs = results[0].probs.data.tolist()

    # Get the predicted class name and the probability
    predicted_class = names_dict[np.argmax(probs)]
    predicted_prob = np.max(probs)

    # Prepare a result dictionary with all class probabilities
    result = {names_dict[i]: probs[i] for i in range(len(probs))}

    return predicted_class, predicted_prob, result

# Gradio interface
inputs = gr.Image(type="filepath")
outputs = [gr.Textbox(label="Predicted Class"), gr.Textbox(label="Confidence"), gr.JSON(label="Class Probabilities")]

# Create a Gradio interface for the function
interface = gr.Interface(fn=classify_image, inputs=inputs, outputs=outputs, title="YOLO Image Classifier")

# Launch the Gradio app
interface.launch()
```
